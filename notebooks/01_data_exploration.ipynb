# Data Exploration
Quick look at CORD, FUNSD, and SROIE datasets to understand what we're working with.

```python
from datasets import load_from_disk
from pathlib import Path
import matplotlib.pyplot as plt
from collections import Counter

DATA_DIR = Path("../data")
```

## CORD Dataset

```python
cord = load_from_disk(str(DATA_DIR / "cord" / "hf_dataset"))
print(f"Splits: {list(cord.keys())}")
print(f"Train: {len(cord['train'])} | Val: {len(cord['validation'])} | Test: {len(cord['test'])}")
print(f"\nFeatures: {cord['train'].features}")
```

```python
# Look at a single example
example = cord['train'][0]
print("Keys:", example.keys())

# Display the image
img = example['image']
plt.figure(figsize=(6, 8))
plt.imshow(img)
plt.title("CORD example receipt")
plt.axis('off')
plt.show()

# Look at the ground truth parse
import json
from ast import literal_eval
gt = literal_eval(example['ground_truth'])
print("\nGround truth parse:")
print(json.dumps(gt['gt_parse'], indent=2))
```

## FUNSD Dataset

```python
funsd = load_from_disk(str(DATA_DIR / "funsd" / "hf_dataset"))
print(f"Splits: {list(funsd.keys())}")
for split in funsd:
    print(f"  {split}: {len(funsd[split])} examples")
print(f"\nFeatures: {funsd['train'].features}")
```

```python
# Look at a FUNSD example
example = funsd['train'][0]
img = example['image']
plt.figure(figsize=(8, 10))
plt.imshow(img)
plt.title("FUNSD example form")
plt.axis('off')
plt.show()

# Check label distribution
all_labels = []
for ex in funsd['train']:
    all_labels.extend(ex['ner_tags'])
label_counts = Counter(all_labels)
print("Label distribution:", label_counts)
```
